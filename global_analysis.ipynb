```python
import pandas as pd
import plotly.express as px
```

```python
# Convert numeric columns
numeric_cols = [
    "Adoption Year",
    "Number of Employees Impacted",
    "New Roles Created",
    "Training Hours Provided",
    "Productivity Change (%)",
    "Employee Sentiment"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
#KPI cards

total_employees = df["Number of Employees Impacted"].sum()
total_roles = df["New Roles Created"].sum()
avg_productivity = df["Productivity Change (%)"].mean()
avg_sentiment = df["Employee Sentiment"].mean()

print("Total Employees Impacted:", total_employees)
print("Total New Roles:", total_roles)
print("Average Productivity Change:", avg_productivity)
print("Average Employee Sentiment:", avg_sentiment)
```

EMPLOYEES IMPACTED BY COUNTRY

```python
#Employees Impacted by Country

country_impact = (
    df.groupby("Country")["Number of Employees Impacted"]
    .sum()
    .reset_index()
)

fig1 = px.choropleth(
    country_impact,
    locations="Country",
    locationmode="country names",
    color="Number of Employees Impacted",
    hover_name="Country",
    color_continuous_scale="Blues",
    title="Global GenAI Workforce Impact"
)

fig1.update_layout(
    template="plotly_white",
    geo=dict(showframe=False, showcoastlines=True)
)

fig1.show()
```

TOP 10 COUNTRIES BY EMPLOYEES IMPACTED

```python
#Top 10 Countries by Employees Impacted

top_countries = country_impact.nlargest(
    10, "Number of Employees Impacted"
).sort_values("Number of Employees Impacted")

fig2 = px.bar(
    top_countries,
    x="Number of Employees Impacted",
    y="Country",
    orientation="h",
    text_auto=".2s",
    title="Top 10 Countries by Employees Impacted"
)

fig2.update_layout(
    template="plotly_white",
    xaxis_title="Employees Impacted",
    yaxis_title="Country"
)

fig2.show()
```

AVERAGE PRODUCTIVITY BY INDUSTRY

```python
#Average Productivity by Industry

industry_productivity = (
    df.groupby("Industry")["Productivity Change (%)"]
    .mean()
    .reset_index()
    .sort_values("Productivity Change (%)", ascending=False)
)

fig3 = px.bar(
    industry_productivity,
    x="Industry",
    y="Productivity Change (%)",
    color="Productivity Change (%)",
    color_continuous_scale="Viridis",
    text_auto=".2f",
    title="Average Productivity Change by Industry"
)

fig3.update_layout(
    template="plotly_white",
    xaxis_title="Industry",
    yaxis_title="Average Productivity Change (%)"
)

fig3.show()
```

NEW ROLES CREATED BY COUNTRY

```python
#New Roles Created by Country

country_roles = (
    df.groupby("Country")["New Roles Created"]
    .sum()
    .reset_index()
    .nlargest(10, "New Roles Created")
    .sort_values("New Roles Created")
)

fig4 = px.bar(
    country_roles,
    x="New Roles Created",
    y="Country",
    orientation="h",
    text_auto=".2s",
    title="Top 10 Countries by New Roles Created"
)

fig4.update_layout(
    template="plotly_white",
    xaxis_title="New Roles Created",
    yaxis_title="Country"
)

fig4.show()
```

COUNTRY IMPACT VS PRODUCTIVITY

```python
#Country Impact vs Productivity

country_summary = (
    df.groupby("Country")
    .agg(
        Employees_Impacted=("Number of Employees Impacted", "sum"),
        Avg_Productivity=("Productivity Change (%)", "mean"),
        New_Roles=("New Roles Created", "sum")
    )
    .reset_index()
)

fig5 = px.scatter(
    country_summary,
    x="Employees_Impacted",
    y="Avg_Productivity",
    size="New_Roles",
    color="Country",
    hover_name="Country",
    size_max=45,
    title="Country-wise Workforce Impact vs Productivity"
)

fig5.update_layout(
    template="plotly_white",
    xaxis_title="Employees Impacted",
    yaxis_title="Average Productivity Change (%)"
)

fig5.show()
```

INDUSTRY- GRNAI TOOL

```python
#Industry → GenAI Tool

treemap_data = (
    df.groupby(["Industry", "GenAI Tool"])
    ["Number of Employees Impacted"]
    .sum()
    .reset_index()
)

fig6 = px.treemap(
    treemap_data,
    path=["Industry", "GenAI Tool"],
    values="Number of Employees Impacted",
    color="Number of Employees Impacted",
    color_continuous_scale="Teal",
    title="Industry and GenAI Tool Distribution"
)

fig6.update_layout(template="plotly_white")
fig6.show()
```

COUNTRY VS GENAI TOOL

```python
#Country vs GenAI Tool

heatmap_data = (
    df.pivot_table(
        index="Country",
        columns="GenAI Tool",
        values="Number of Employees Impacted",
        aggfunc="sum",
        fill_value=0
    )
)

# Top 15 countries by total employee impact
top15 = heatmap_data.sum(axis=1).nlargest(15).index
heatmap_data = heatmap_data.loc[top15]

fig7 = px.imshow(
    heatmap_data,
    aspect="auto",
    color_continuous_scale="Blues",
    labels={
        "x": "GenAI Tool",
        "y": "Country",
        "color": "Employees Impacted"
    },
    title="Country vs GenAI Tool — Workforce Impact"
)

fig7.update_layout(template="plotly_white")
fig7.show()
```
