```python
import pandas as pd
import plotly.express as px
```

```python
# Convert numeric columns
numeric_cols = [
    "Adoption Year",
    "Number of Employees Impacted",
    "New Roles Created",
    "Training Hours Provided",
    "Productivity Change (%)",
    "Employee Sentiment"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
#KPI cards

total_employees = df["Number of Employees Impacted"].sum()
total_roles = df["New Roles Created"].sum()
avg_productivity = df["Productivity Change (%)"].mean()
avg_sentiment = df["Employee Sentiment"].mean()

print("Total Employees Impacted:", total_employees)
print("Total New Roles:", total_roles)
print("Average Productivity Change:", avg_productivity)
print("Average Employee Sentiment:", avg_sentiment)
```

```python
#GenAI Tool Adoption Count

tool_adoption = (
    df["GenAI Tool"]
    .value_counts()
    .reset_index()
)

tool_adoption.columns = ["GenAI Tool", "Adoption Count"]

fig1 = px.bar(
    tool_adoption,
    x="GenAI Tool",
    y="Adoption Count",
    text_auto=True,
    title="GenAI Tool Adoption Count"
)

fig1.update_layout(
    xaxis_title="GenAI Tool",
    yaxis_title="Number of Records",
    template="plotly_white"
)

fig1.show()
```

```python
#employees Impacted by GenAI Tool

employees_tool = (
    df.groupby("GenAI Tool")["Number of Employees Impacted"]
    .sum()
    .reset_index()
    .sort_values(
        "Number of Employees Impacted",
        ascending=False
    )
)

fig2 = px.bar(
    employees_tool,
    x="GenAI Tool",
    y="Number of Employees Impacted",
    text_auto=".2s",
    title="Employees Impacted by GenAI Tool"
)

fig2.update_layout(
    xaxis_title="GenAI Tool",
    yaxis_title="Employees Impacted",
    template="plotly_white"
)

fig2.show()
```

```python
#Average Productivity Change by GenAI Tool

productivity_tool = (
    df.groupby("GenAI Tool")["Productivity Change (%)"]
    .mean()
    .reset_index()
    .sort_values(
        "Productivity Change (%)",
        ascending=False
    )
)

fig3 = px.bar(
    productivity_tool,
    x="GenAI Tool",
    y="Productivity Change (%)",
    text_auto=".2f",
    title="Average Productivity Change by GenAI Tool"
)

fig3.update_layout(
    xaxis_title="GenAI Tool",
    yaxis_title="Average Productivity Change (%)",
    template="plotly_white"
)

fig3.show()
```

```python
#Average Training Hours by Tool

training_tool = (
    df.groupby("GenAI Tool")["Training Hours Provided"]
    .mean()
    .reset_index()
    .sort_values(
        "Training Hours Provided",
        ascending=False
    )
)

fig4 = px.bar(
    training_tool,
    x="GenAI Tool",
    y="Training Hours Provided",
    text_auto=".2f",
    title="Average Training Hours by GenAI Tool"
)

fig4.update_layout(
    xaxis_title="GenAI Tool",
    yaxis_title="Average Training Hours",
    template="plotly_white"
)

fig4.show()
```

```python
#Tool Adoption Percentage

fig5 = px.pie(
    tool_adoption,
    names="GenAI Tool",
    values="Adoption Count",
    hole=0.55,
    title="GenAI Tool Adoption Percentage"
)

fig5.update_traces(
    textposition="inside",
    textinfo="percent+label"
)

fig5.update_layout(
    template="plotly_white"
)

fig5.show()
```

```python
#Training Hours vs Productivity Change
tool_scatter = (
    df.groupby("GenAI Tool")
    .agg(
        Avg_Training_Hours=("Training Hours Provided", "mean"),
        Avg_Productivity=("Productivity Change (%)", "mean"),
        Employees_Impacted=("Number of Employees Impacted", "sum")
    )
    .reset_index()
)

fig6 = px.scatter(
    tool_scatter,
    x="Avg_Training_Hours",
    y="Avg_Productivity",
    size="Employees_Impacted",
    color="GenAI Tool",
    hover_name="GenAI Tool",
    title="Training vs Productivity by GenAI Tool",
    size_max=50
)

fig6.update_layout(
    xaxis_title="Average Training Hours",
    yaxis_title="Average Productivity Change (%)",
    template="plotly_white"
)

fig6.show()
```
