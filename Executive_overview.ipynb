```python
import pandas as pd
import plotly.express as px
```

```python
# Convert numeric columns
numeric_cols = [
    "Adoption Year",
    "Number of Employees Impacted",
    "New Roles Created",
    "Training Hours Provided",
    "Productivity Change (%)",
    "Employee Sentiment"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
#KPI cards

total_employees = df["Number of Employees Impacted"].sum()
total_roles = df["New Roles Created"].sum()
avg_productivity = df["Productivity Change (%)"].mean()
avg_sentiment = df["Employee Sentiment"].mean()

print("Total Employees Impacted:", total_employees)
print("Total New Roles:", total_roles)
print("Average Productivity Change:", avg_productivity)
print("Average Employee Sentiment:", avg_sentiment)
```

```python
#GenAI Adoption by Industry

industry_adoption = (
    df.groupby("Industry")["Number of Employees Impacted"]
    .sum()
    .reset_index()
    .sort_values("Number of Employees Impacted", ascending=False)
)

fig1 = px.bar(
    industry_adoption,
    x="Industry",
    y="Number of Employees Impacted",
    title="GenAI Impact by Industry",
    text_auto=".2s"
)

fig1.update_layout(
    xaxis_title="Industry",
    yaxis_title="Employees Impacted",
    template="plotly_white"
)

fig1.show()
```

```python
#Employees impacted by GenAI Tool

tool_impact = (
    df.groupby("GenAI Tool")["Number of Employees Impacted"]
    .sum()
    .reset_index()
    .sort_values("Number of Employees Impacted", ascending=False)
)

fig2 = px.bar(
    tool_impact,
    x="GenAI Tool",
    y="Number of Employees Impacted",
    title="Employees Impacted by GenAI Tool",
    text_auto=".2s"
)

fig2.update_layout(
    xaxis_title="GenAI Tool",
    yaxis_title="Employees Impacted",
    template="plotly_white"
)

fig2.show()
```

```python
#GenAI Adoption Trend Over Years

year_trend = (
    df.groupby("Adoption Year")["Number of Employees Impacted"]
    .sum()
    .reset_index()
    .sort_values("Adoption Year")
)

fig3 = px.line(
    year_trend,
    x="Adoption Year",
    y="Number of Employees Impacted",
    markers=True,
    title="GenAI Adoption Trend Over Years"
)

fig3.update_layout(
    xaxis_title="Adoption Year",
    yaxis_title="Employees Impacted",
    template="plotly_white"
)

fig3.show()
```

```python
#Employee Sentiment Distribution
    
fig4 = px.pie(
    tool_impact,
    names="GenAI Tool",
    values="Number of Employees Impacted",
    hole=0.55,
    title="Share of Employees Impacted by GenAI Tool"
)

fig4.update_traces(
    textposition="inside",
    textinfo="percent+label"
)

fig4.update_layout(
    template="plotly_white"
)

fig4.show()
```

```python
#roles created by industry

industry_roles = (
    df.groupby("Industry")
    ["New Roles Created"]
    .sum()
    .reset_index()
    .sort_values("New Roles Created", ascending=False)
)

fig = px.bar(
    industry_roles,
    x="Industry",
    y="New Roles Created",
    text_auto=".2s",
    title="New Roles Created by Industry"
)

fig.update_layout(
    xaxis_title="Industry",
    yaxis_title="New Roles Created",
    template="plotly_white"
)

fig.show()
```
