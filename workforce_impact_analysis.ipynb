```python
import pandas as pd
import plotly.express as px
```

```python
# Convert numeric columns
numeric_cols = [
    "Adoption Year",
    "Number of Employees Impacted",
    "New Roles Created",
    "Training Hours Provided",
    "Productivity Change (%)",
    "Employee Sentiment"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
#KPI cards

total_employees = df["Number of Employees Impacted"].sum()
total_roles = df["New Roles Created"].sum()
avg_productivity = df["Productivity Change (%)"].mean()
avg_sentiment = df["Employee Sentiment"].mean()

print("Total Employees Impacted:", total_employees)
print("Total New Roles:", total_roles)
print("Average Productivity Change:", avg_productivity)
print("Average Employee Sentiment:", avg_sentiment)
```

```python
#Employees impacted by industry

employees_industry = (
    df.groupby("Industry")["Number of Employees Impacted"]
    .sum()
    .reset_index()
    .sort_values("Number of Employees Impacted", ascending=False)
)

fig1 = px.bar(
    employees_industry,
    x="Industry",
    y="Number of Employees Impacted",
    color="Industry",
    text_auto=".2s",
    title="Employees Impacted by Industry"
)

fig1.update_layout(
    template="plotly_white",
    showlegend=False,
    xaxis_title="Industry",
    yaxis_title="Employees Impacted"
)

fig1.show()
```

```python
#New Roles Created by Industry
roles_industry = (
    df.groupby("Industry")["New Roles Created"]
    .sum()
    .reset_index()
    .sort_values("New Roles Created", ascending=False)
)

fig2 = px.bar(
    roles_industry,
    x="Industry",
    y="New Roles Created",
    color="Industry",
    text_auto=".2s",
    title="New Roles Created by Industry"
)

fig2.update_layout(
    template="plotly_white",
    showlegend=False,
    xaxis_title="Industry",
    yaxis_title="New Roles Created"
)

fig2.show()
```

```python
#Employees Impacted vs New Roles Created

fig3 = px.scatter(
    employees_industry.merge(roles_industry, on="Industry"),
    x="Number of Employees Impacted",
    y="New Roles Created",
    size="Number of Employees Impacted",
    color="Industry",
    hover_name="Industry",
    title="Employees Impacted vs New Roles Created"
)

fig3.update_layout(
    template="plotly_white",
    xaxis_title="Employees Impacted",
    yaxis_title="New Roles Created"
)

fig3.show()
```

```python
#Industry Workforce Impact

workforce_summary = (
    df.groupby("Industry")
    .agg(
        Employees_Impacted=("Number of Employees Impacted", "sum"),
        New_Roles=("New Roles Created", "sum"),
        Avg_Productivity=("Productivity Change (%)", "mean")
    )
    .reset_index()
)

fig4 = px.scatter(
    workforce_summary,
    x="Employees_Impacted",
    y="New_Roles",
    size="Employees_Impacted",
    color="Avg_Productivity",
    hover_name="Industry",
    size_max=55,
    title="Industry Workforce Impact Overview",
    color_continuous_scale="Viridis"
)

fig4.update_layout(
    template="plotly_white",
    xaxis_title="Employees Impacted",
    yaxis_title="New Roles Created",
    coloraxis_colorbar_title="Avg Productivity (%)"
)

fig4.show()
```

```python
#Training Hours by Industry

training_industry = (
    df.groupby("Industry")["Training Hours Provided"]
    .mean()
    .reset_index()
    .sort_values("Training Hours Provided", ascending=False)
)

fig5 = px.bar(
    training_industry,
    x="Industry",
    y="Training Hours Provided",
    color="Industry",
    text_auto=".2f",
    title="Average Training Hours by Industry"
)

fig5.update_layout(
    template="plotly_white",
    showlegend=False,
    xaxis_title="Industry",
    yaxis_title="Average Training Hours"
)

fig5.show()
```

```python
#Training Hours vs New Roles

training_roles = (
    df.groupby("Industry")
    .agg(
        Avg_Training_Hours=("Training Hours Provided", "mean"),
        Total_New_Roles=("New Roles Created", "sum")
    )
    .reset_index()
)

fig6 = px.scatter(
    training_roles,
    x="Avg_Training_Hours",
    y="Total_New_Roles",
    color="Industry",
    hover_name="Industry",
    title="Training Hours vs New Roles Created"
)

fig6.update_layout(
    template="plotly_white",
    xaxis_title="Average Training Hours",
    yaxis_title="Total New Roles Created"
)

fig6.show()
```

```python
#Industry Workforce Transformation
treemap_data = (
    df.groupby(["Industry", "GenAI Tool"])["New Roles Created"]
    .sum()
    .reset_index()
)

fig7 = px.treemap(
    treemap_data,
    path=["Industry", "GenAI Tool"],
    values="New Roles Created",
    color="New Roles Created",
    color_continuous_scale="Blues",
    title="Workforce Transformation by Industry and GenAI Tool"
)

fig7.update_layout(template="plotly_white")

fig7.show()
```
